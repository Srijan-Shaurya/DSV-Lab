```python
import pandas as pd

sales = pd.Series({'Laptop': 450, 'Mobile': 600, 'Tablet': 300, 'TV': 750})

print("Values:", sales.values)
print("Index:", sales.index)
print("Mobile Sales:", sales['Mobile'])
print("Sales > 500:\n", sales[sales > 500])
print("Is Laptop in series?:", 'Laptop' in sales)

sales_dict = sales.to_dict()
print("Dictionary:", sales_dict)
```
